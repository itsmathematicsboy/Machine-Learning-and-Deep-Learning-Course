```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neural_network import MLPRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import RobustScaler
from sklearn.metrics import mean_absolute_percentage_error, root_mean_squared_error
```

```python
path = r'https://github.com/itsmathematicsboy/Machine-Learning-and-Deep-Learning-Course/raw/main/Machine%20Learning/Dataset/tvmarketing.csv'
df = pd.read_csv(path)
```

```python
df.head()
```

```python
x = df['TV'].values.reshape(-1, 1)
y = df['Sales'].values.reshape(-1, 1)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, train_size = .7, test_size = .3)
```

```python
robust = RobustScaler()
```

```python
robust_x = robust
robust_y = robust
```

```python
x_train_scaler = robust_x.fit_transform(x_train)
x_test_scaler = robust_x.transform(x_test)
y_train_scaler = robust_y.fit_transform(y_train)
y_test_scaler = robust_y.transform(y_test)
```

```python
mlp = MLPRegressor(activation = 'relu',
                  solver = 'sgd', verbose = True)
```

```python
model = mlp.fit(x_train_scaler, y_train_scaler)
```

```python
y_pred = model.predict(x_test_scaler)
```

```python
def mape(y_pred, y_test):
    return mean_absolute_percentage_error(y_pred, y_test)

def rmse(y_pred, y_test):
    return root_mean_squared_error(y_pred, y_test)
```

```python
print('MAPE:', mape(y_pred, y_test_scaler))
print('RMSE:', rmse(y_pred, y_test_scaler))
```
