```python
# Importing Library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import Normalizer
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Importing Dataset
df = pd.read_csv('tvmarketing.csv')
```

```python
# Initialize the Feature and Target Variable
x = df['TV'].values.reshape(-1, 1)
y = df['Sales'].values.reshape(-1, 1)
```

```python
# Train Test Splitting
x_train, x_test, y_train, y_test = train_test_split(x, y, train_size = .7, test_size = .3)
```

```python
# Initialize Normalizer into norm variable
norm = Normalizer()
```

```python
# initialize Normalizer for feature & Target
norm_x = norm
norm_y = norm
```

```python
# Scaling train and test feature and target
x_train_scaler = norm_x.fit_transform(x_train)
x_test_scaler = norm_x.transform(x_test)
y_train_scaler = norm_y.fit_transform(y_train)
y_test_scaler = norm_y.transform(y_test)
```

```python
# Initialize model with 5 nearest neighbors
knn = KNeighborsRegressor(n_neighbors = 5)
```

```python
# fit and training the train data that already scaled before
model = knn.fit(x_train_scaler, y_train_scaler)
```

```python
# testing the test data
y_pred = model.predict(x_test_scaler)
```

```python
# function for metrics
def mae(y_pred, y_test):
    '''
    Compute the mean absolute error with predict and testing data
    '''
    return mean_absolute_error(y_pred, y_test)

def mse(y_pred, y_test):
    '''
    Compute the mean squared error with predict and testing data
    '''
    return mean_squared_error(y_pred, y_test)

def r2(y_pred, y_test):
    '''
    Compute the r2 score
    '''
    return r2_score(y_pred, y_test)
```

```python
# Result of testing
print('Mean Absolute Error:', mae(y_pred, y_test_scaler))
print('Mean Squared Error:', mse(y_pred, y_test_scaler))
print('R2 Score:', r2(y_pred, y_test_scaler))
```
